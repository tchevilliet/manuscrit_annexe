```python
# bat_db_dict = bd.Database("2026-4-22-16h7_buildings_cons_RE2020").load()

n = 10

indicators = df_tirages_composants["indicator"].unique().to_list()
tirages = list(df_tirages_composants["iteration"].unique())[:n]

df_filtered = df_tirages_composants.filter(
    pl.col("indicator").is_in(indicators) & pl.col("iteration").is_in(tirages)
)
# On construit un polars dataframe des composants par bat avec infos nécessaires au calcul du score
rows_comp = []
for bat_node in tqdm(bat_db_dict.values()):
    flux_ref = next(e["amount"] for e in bat_node["exchanges"] if e["type"] == "production")
    for e in bat_node["exchanges"]:
        if e["type"] == "production":
            continue
        k = e["input"][1]
        if "consommations_reseau" in k:
            k = "consommations_reseau"
        rows_comp.append({"batiment": bat_node["name"], "usage": bat_node["usage"],
                           "composant": k, "ratio": e["amount"] / flux_ref})


# en sommant les même composant (des différents lots par ex) 
df_comp_bat = (
    pl.DataFrame(rows_comp)
    .group_by(["batiment", "usage", "composant"])
    .agg(pl.col("ratio").sum())
) 

# on fait les opérations avec polars : rapide a priori

Hb = (
    df_comp_bat.join(
        df_filtered, left_on="composant", right_on="demand", how="inner"
        )# On joitn le df initial 
    .with_columns(
        # ajout d'une colonne produit du ratio de cahque comp par bat au score du comp
        (pl.col("ratio") * pl.col("score")).alias("contrib")
        )
    .group_by(["iteration", "indicator", "batiment", "usage"])
    .agg(pl.col("contrib").sum().alias("score_final")) # on somme par bat pour chaque tirage et indicateur
    .select(["iteration", "batiment", "usage", "indicator", "score_final"])
)

# rows = []

# df_filtered = df_tirages_composants.filter(pl.col("indicator") == indicator)

# # On calcule toutes les infos nécessaires sur chaque bat
# bat_info = []




# for i in tqdm(tirages[:n],desc = "sur tirages", total = n) :
#     tmp = df_tirages_composants.filter(pl.col("iteration")==i)
#     d = tmp.to_dict(as_series=False)
#     dict_scores = dict(zip(d["demand"], d["score"]))

#     for bat_node in tqdm(bat_db_dict.values(),desc = "sur bâtiments",total = len(bat_db_dict)) :
#         score = 0
#         # 1 tout le temps, mais bonne pratique
#         flux_ref = [
#             e for e in bat_node["exchanges"] if e["type"]=="production"
#             ][0]["amount"] 
#         edges = [e for e in bat_node["exchanges"] if e["type"]!="production"]
#         for e in edges :
#             k = e["input"][1]
#             if "consommations_reseau" in k :
#                 k = "consommations_reseau"
#             score_comp = dict_scores[k]
#             score += e["amount"]*score_comp/flux_ref

#         rows.append({
#             "iteration" : i,
#             "batiment" : bat_node["name"],
#             "usage" : bat_node["usage"],
#             "indicator" : indicator,
#             "score" : score
#         })
    
# Hb = pl.DataFrame(
#     rows,
#     schema={
#         "iteration": pl.Int64,
#         "batiment": pl.String,
#         "usage": pl.String,
#         "indicator": pl.String,
#         "score": pl.Float64
#         }
#     )

```

```python
import numpy as np
import scipy.sparse as sp
import polars as pl
import gc

n=10

output_dir = "/home/thibault.chevilliet@enpc.fr/Documents/2601_archetypes_re2020/data/processed/Hb"

# ============ 1. PRÉCALCUL UNE SEULE FOIS (indépendant indicateur/tirage) ============

all_keys = set()
bat_edges = []  # (nom, usage, {k: ratio_cumulé})

for bat_node in tqdm(bat_db_dict.values(), desc="précalcul bâtiments", total=len(bat_db_dict)):
    flux_ref = next(e["amount"] for e in bat_node["exchanges"] if e["type"] == "production")
    edges = {}
    for e in bat_node["exchanges"]:
        if e["type"] == "production":
            continue
        k = e["input"][1]
        if "consommations_reseau" in k:
            k = "consommations_reseau"
        edges[k] = edges.get(k, 0.0) + e["amount"] / flux_ref  # cumule si clé répétée
        all_keys.add(k)
    bat_edges.append((bat_node["name"], bat_node["usage"], edges))

key_list = sorted(all_keys)
key_to_idx = {k: i for i, k in enumerate(key_list)}
n_bat, n_keys = len(bat_edges), len(key_list)

rows_idx, cols_idx, vals = [], [], []
noms_bat = np.empty(n_bat, dtype=object)
usages_bat = np.empty(n_bat, dtype=object)
for i, (nom, usage, edges) in enumerate(bat_edges):
    noms_bat[i], usages_bat[i] = nom, usage
    for k, ratio in edges.items():
        rows_idx.append(i); cols_idx.append(key_to_idx[k]); vals.append(ratio)

# matrice creuse (n_bat x n_keys) — quasi rien en RAM (nnz ~ n_bat * nb_edges_moyen)
A = sp.csr_matrix((vals, (rows_idx, cols_idx)), shape=(n_bat, n_keys), dtype=np.float32)
print(f"A : {n_bat} bâtiments x {n_keys} clés, {A.nnz} non-zéros, {A.data.nbytes/1e6:.1f} Mo")

# ============ 2. BOUCLE INDICATEUR PAR INDICATEUR ============

indicators = df_tirages_composants["indicator"].unique().to_list()
tirages = sorted(df_tirages_composants["iteration"].unique().to_list())[:n]

for indicator in tqdm(indicators, desc="indicateurs"):
    df_ind = df_tirages_composants.filter(
        (pl.col("indicator") == indicator) & pl.col("iteration").is_in(tirages)
    ).select(["demand", "iteration", "score"])

    pivot = df_ind.pivot(values="score", index="demand", on="iteration")
    demand_order = pivot["demand"].to_list()
    iter_cols = [c for c in pivot.columns if c != "demand"]
    tirages_used = [int(c) for c in iter_cols]

    pivot_vals = pivot.select(iter_cols).to_numpy().astype(np.float32)  # (n_demand, n_tirages)

    # aligne sur le vocabulaire global key_list (clés absentes -> 0)
    S = np.zeros((n_keys, len(iter_cols)), dtype=np.float32)
    missing = []
    for i, k in enumerate(demand_order):
        if k in key_to_idx:
            S[key_to_idx[k], :] = pivot_vals[i, :]
        else:
            missing.append(k)
    if missing:
        print(f"⚠ {indicator}: clés non utilisées par aucun bâtiment: {missing[:5]}...")

    # calcul vectorisé : (n_bat x n_keys) @ (n_keys x n_tirages) -> (n_bat x n_tirages)
    scores = A @ S   # dense, ~320 Mo en float32, libéré en fin de boucle

    # ---- a) juste tracer la distribution (le moins coûteux) ----
    # plt.figure(); plt.hist(scores.ravel(), bins=200); plt.title(indicator); plt.savefig(...)

    # ---- b) si tu veux aussi garder le détail sur disque (1 fichier/indicateur) ----
    df_out = pl.DataFrame({
        "iteration": np.tile(np.array(tirages_used, dtype=np.int64), n_bat),
        "batiment": pl.Series(np.repeat(noms_bat, len(tirages_used))).cast(pl.Categorical),
        "usage": pl.Series(np.repeat(usages_bat, len(tirages_used))).cast(pl.Categorical),
        "score": scores.ravel(),
    }).with_columns(pl.lit(indicator).alias("indicator"))
    df_out.write_parquet(f"{output_dir}/scores_{indicator}.parquet")

    del scores, S, df_out, pivot, pivot_vals
    gc.collect()
```
